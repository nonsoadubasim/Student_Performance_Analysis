Student name

```python
Students = ["John", "Mary", "Peter", "Grace", "David"]

Scores = [75, 90, 65, 85,35]

print("STUDENTS PERFORMANCE DATA ANALYSIS")
print("---------------------------------")

print("Students:", Students)
print("Scores:", Scores)
```

```python
# Calculate the average score

Total_Scores = sum(Scores)

Average_Score = Total_Scores / len(Scores)

print("Total Score:", Total_Scores)
print("Average Score:", Average_Score)
```

```python
# Find the highest and lowest scores

Highest_scores = max(Scores)
Lowest_scores = min(Scores)

Highest_students = Students[Scores.index(Highest_scores)]
Lowest_students = Students[Scores.index(Lowest_scores)]

print("Highest scores:", Highest_scores)
print("Top student:", Highest_students)

print("Lowest scores:", Lowest_scores)
print("Lowest performing student:", Lowest_students)
```

```python
# Classify student as Pass or fail

for i in range(len(Students)):
    if Scores[i] >= 50:
        print(Students[i], "-", "PASS")
    else:
        print(Students[i], "-", "FAIL")
```

```python
# Calculate pass rate

pass_count = 0
fail_count = 0

for score in Scores:
    if score >=50:
        pass_count += 1
    else:
        fail_count += 1
        
pass_percentage = (pass_count / len(Scores)) * 100

print("Number of Students:", len(Students))
print("Student who Passed:", pass_count)
print("Student who failed:", fail_count)
print("Pass percentage:", pass_percentage, "%")        
```

```python
# Calculate the average score
average_score = sum(Scores) / len(Scores)

# Create the chart
plt.figure(figsize=(8, 5))

plt.bar(Students, Scores)

plt.bar(Students, Scores, color="purple", edgecolor="black")

# Display each student's score
for i in range(len(Students)):
    plt.text(i, Scores[i] + 1, str(Scores[i]), ha="center")

# Draw a line showing the average    
plt.axhline(

    y=average_score,

    color="Black",

    linestyle="--",

    label=f"Average Score: {average_score:.1f}")    

plt.title("Students_Performance_Analysis")
plt.xlabel("Students")
plt.ylabel("scores")

plt.ylim(0, 100)

plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python
# Pie chart showing pass and fail percentages

labels = ["Passed", "Failed"]
results = [pass_count, fail_count]

plt.figure(figsize=(7, 7))

plt.pie(
    results,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Pass and Fail Percentage")

plt.show()
```

```python
import pandas as pd 
```

```python
data = {
    "Students": ["John", "Mary", "Peter", "Grace", "David"],
    "Scores": [75, 90, 65, 85, 35]
}

df = pd.DataFrame(data)
print(df)
```

```python
results = []

for Scores in df["Scores"]:
    if Scores >= 50:
        results.append("PASS")
    else:
        results.append("FAIL")

df["Results"] = results

print(df)
```

```python
print("Average Scores:", df["Scores"].mean())

print("Highest Scores:", df["Scores"].max())

print("Lowest Scores:", df["Scores"].min())

print("Total Students:", df["Students"].count())
```

Analyzing Student Performance using pandas


```python
def performance(scores):
    if scores >= 80:
        return "Excellent" 
    elif scores >= 70:
        return "good" 
    elif scores >= 50:
        return "Average" 
    else:
        return "Need improvement" 
    
df["Performance"] = df["Scores"].apply(performance)

df["Result"] = df["Scores"].apply(
    lambda score: "Pass" 
    if score >= 50
    else "fail"
)
     

print(df)  
```

```python
print(df["Result"].value_counts())
```

```python
print(df["Performance"].value_counts())
```

Calculate the Average score each performance category

```python
average_by_performance = df.groupby("Performance")["Scores"].mean()
print (average_by_performance)
```

Find the student with high score 

```python
top_student = df.loc[df["Scores"].idxmax()]

print("Top Students:", top_student["Students"])
print("Highest Scores:", top_student["Scores"])
```

Find the student with lowest score 

```python
Lowest_index =df["Scores"].idxmin()

Lowest_students = df.loc[Lowest_index]

print("Student Needing impovement:", Lowest_students["Students"])
print("Lowest Scores:", Lowest_students ["Scores"])
```

Find Students who need improvement

```python
Students_needing_improvement = df[df["Scores"] < 50]

print("Students who Needing Improvement:")
print(Students_needing_improvement)
```

Find student who has the highest score 

```python
Highest_students = df.loc[df["Scores"] == df["Scores"].max()]

print("Highest scoring students:")
print(Highest_students)
```

Calculate the percentage of student who passed


```python
passed_students = (df["Results"] == "PASS").sum()

total_students = len(df)

pass_percentage =(passed_students / total_students) * 100

print("Number of Students Who Passed:", passed_students)
print("Total Number of student:", total_students)
print("Pass Percentage:", pass_percentage, "%")
```

Calculate the failure percentage

```python
failed_students = (df["Results"] == "FAIL").sum()

total_students = len(df)

failure_percentage = (failed_students / total_students) * 100

print("Number of student who failed:", failed_students)
print("Total number of student:", total_students)
print("failure percentage:", failure_percentage, "%")
```

Display a summary of the students'performance

```python
print("==== STUDENTS PERFORMANCE REPORT ===")
print("total_student:", len(df))
print("Highest scores:", df["Scores"].max())
print("lowest scores:", df["Scores"].min())
print("Average score:", df["Scores"].mean())
print("pass percentage:", pass_percentage, "%")
print("failure percentage:", failure_percentage, "%")
print("=====================================")
```

```python
plt.Figure(figsize=(15, 8))

plt.bar(df["Students"], df["Scores"])

plt.bar(df["Students"], df["Scores"], color="green", edgecolor="red")

plt.title("Students' Performance", fontsize=20)
plt.xlabel("Students", fontsize=15)
plt.ylabel("Scores", fontsize=15)


plt.ylim(0, 100)

plt.grid(axis="y", linestyle="--", alpha=0.5)

for i in range(len(df)):
    plt.text(i, df["Scores"].iloc[i] + 1,
             str(df["Scores"].iloc[i]),
             ha="center")
    
plt.savefig("students_performance.png", dpi=300, bbox_inches="tight")
plt.show()
```
